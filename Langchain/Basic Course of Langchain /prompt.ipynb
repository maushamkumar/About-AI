## Prompt Templates

- Prompt templates help to translate user input and parameters into instructions for a language model 

- Prompt Templates take as input a dictionary, where each key represents a variable in the prompt template to fill in.

- Prompt Templates output a PromptValue. This PromptValue can be passed to an LLM or a ChatModel, and can also be cast to a string or a list of messages. The reason this PromptValue exists is to make it easy to switch between strings and messages.

- There are a few different types of prompt templates:

## ChatPromptTemplates

- These prompt templates are used to format a list of messages 

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt_template = ChatPromptTemplate([
    ("system", "You are a helpful assistant"),
    ("user", "Tell me a joke about {topic}")
])
```

```python
prompt_template.invoke({"topic": "cats"})
```

- In the above code example, this ChatPromptTemplate will construct two message when called. 
- The first is system message, that has no variables to format. 
- HumanMessage will be formatted by the `topic` variable the user passes in. 

```python
prompt_template.messages[0].prompt.template
```

```python
prompt = prompt_template.invoke({"topic": "cats"})
```

## MessagesPlaceholder

- This prompt template is responsible for adding a list of messages in a particular place.
 
- In the above ChatPromptTemplate, we saw how we could format two messages, each one a string. But what if we wanted the user to pass in a list of messages that we would slot into a particular spot? This is how you use MessagesPlaceholder.



```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage
```

```python
prompt_template = ChatPromptTemplate([
    ('system', 'you are helpful assistant'), 
    MessagesPlaceholder('msgs')
])
```

```python
prompt_template.invoke({
    'msgs':[HumanMessage(content="hi!")]
})
```

```python
# More complex example with conversation history
messages_to_pass = [
    HumanMessage(content="What's the capital of France?"),
    AIMessage(content="The capital of France is Paris."),
    HumanMessage(content="And what about Germany?")
]

formatted_prompt = prompt_template.invoke({"msgs": messages_to_pass})
print(formatted_prompt)
```
